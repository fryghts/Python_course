## 1. Работа с пропущенными данными (NaN)

### Что такое NaN?
**NaN** (Not a Number) — специальное значение для отсутствующих данных.

### Причины появления пропусков:
- Данные не были собраны
- Ошибка при вводе
- Датчик не работал
- Просто нет информации

### Методы для работы с NaN:
- `.isnull()` / `.isna()` — найти пропуски (True/False)
- `.notnull()` / `.notna()` — найти НЕ пропуски
- `.fillna()` — заполнить пропуски
- `.dropna()` — удалить строки/столбцы с пропусками

## 3. Применение функций

### Методы для применения функций:
- `.apply()` — применить функцию к колонке или строке
- `.map()` — применить функцию/словарь к Series
- `.applymap()` — применить функцию к каждому элементу (устарело, используйте .map())

### Когда что использовать?
- **Векторизованные операции** (быстрее всего): `df['col'] * 2`
- **.apply()**: когда нужна сложная логика
- **.map()**: для замены значений или простых функций

## 4. Изменение типов данных

- `int64` — целые числа
