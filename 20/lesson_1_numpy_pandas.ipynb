## 1. Введение в NumPy

### Что такое NumPy?

**NumPy** (Numerical Python) — это библиотека для работы с многомерными массивами и математических вычислений.

### Зачем нужен NumPy?

- **Скорость**: операции с NumPy в десятки раз быстрее, чем с обычными списками Python
- **Удобство**: простой синтаксис для сложных математических операций
- **Функциональность**: огромное количество встроенных функций

### NumPy vs Python списки:

```python
# Python список
python_list = [1, 2, 3, 4, 5]
doubled = [x * 2 for x in python_list]  # Нужен цикл

# NumPy array
numpy_array = np.array([1, 2, 3, 4, 5])
doubled = numpy_array * 2  # Операция сразу для всех элементов
```

## 4. Введение в Pandas

### Что такое Pandas?

**Pandas** — это библиотека для анализа и обработки структурированных данных (таблиц).

### Две основные структуры данных:

1. **Series** — одномерный массив с индексами (похож на столбец таблицы)
2. **DataFrame** — двумерная таблица данных (как Excel таблица)

### Почему Pandas?

- Работа с табличными данными
- Легкая фильтрация и группировка
- Чтение/запись различных форматов (CSV, Excel, SQL)
- Инструменты для очистки данных
- Встроенная визуализация
