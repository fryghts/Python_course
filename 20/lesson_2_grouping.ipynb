## 1. Группировка данных (.groupby())

### Концепция Split-Apply-Combine

**GroupBy** работает в 3 этапа:
1. **Split** — разделить данные на группы
2. **Apply** — применить функцию к каждой группе
3. **Combine** — объединить результаты

### Синтаксис:
```python
df.groupby('column').agg_function()
```

### Зачем нужна группировка?
- Найти среднее по категориям

## 5. Объединение данных

### Типы объединения:

1. **concat()** — склеивание DataFrame'ов вдоль оси
2. **merge()** — SQL-style join по ключевым колонкам
3. **join()** — объединение по индексу

### Типы merge (как в SQL):
- `inner` — только совпадающие записи
- `left` — все из левого + совпадающие из правого
- `right` — все из правого + совпадающие из левого
- `outer` — все записи из обоих
