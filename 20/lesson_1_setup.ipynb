## 1. Виртуальные окружения (venv)

### Что такое виртуальное окружение?

Виртуальное окружение — это изолированная среда для Python-проекта, которая позволяет:
- Устанавливать библиотеки только для конкретного проекта
- Избегать конфликтов версий между разными проектами
- Легко делиться списком зависимостей с другими разработчиками

### Как создать виртуальное окружение?

**Windows:**
```bash
python -m venv my_venv
my_venv\Scripts\activate
```

**macOS/Linux:**
```bash
python3 -m venv my_venv
source my_venv/bin/activate
```

После активации вы увидите `(my_venv)` в начале строки терминала.

## 2. Менеджер пакетов pip

### Что такое pip?

`pip` — это менеджер пакетов для Python. С его помощью можно:
- Устанавливать библиотеки из PyPI (Python Package Index)
- Обновлять существующие библиотеки
- Удалять ненужные библиотеки
- Создавать список зависимостей проекта

### Основные команды pip:

```bash
# Установка библиотеки
pip install pandas

# Установка конкретной версии
pip install pandas==2.0.0

# Установка из файла requirements.txt
pip install -r requirements.txt

# Показать установленные библиотеки
pip list

# Информация о конкретной библиотеке
pip show pandas

# Обновление библиотеки
pip install --upgrade pandas
```

### Файл requirements.txt

Это текстовый файл со списком всех библиотек проекта:
```
pandas>=2.0.0
numpy>=1.24.0
matplotlib>=3.7.0
```

## 3. Jupyter Notebook

### Что такое Jupyter Notebook?

Jupyter Notebook — это интерактивная среда для:
- Написания и выполнения кода
- Добавления текстовых пояснений
- Визуализации данных
- Документирования процесса анализа

### Типы ячеек:

1. **Code** — для Python кода (выполняется при нажатии Shift+Enter)
2. **Markdown** — для текста, заголовков, списков (как эта ячейка)

### Полезные горячие клавиши:

- `Shift + Enter` — выполнить ячейку и перейти к следующей
- `Ctrl + Enter` — выполнить ячейку и остаться в ней
- `A` — создать ячейку выше (Above)
- `B` — создать ячейку ниже (Below)
- `D + D` — удалить ячейку
- `M` — превратить ячейку в Markdown
- `Y` — превратить ячейку в Code

### Магические команды:

В Jupyter есть специальные команды, начинающиеся с `%`:
- `%pwd` — показать текущую директорию
- `%ls` — показать файлы в директории
- `%time` — измерить время выполнения команды
- `%matplotlib inline` — отображать графики прямо в notebook
